# 02 · Body proportions and missing hips
Use the same MediaPipe model, visibility threshold, and pixel-distance calculation as Drape. Compare one complete photo with a crop of the same photo that ends above the hips. A returned ratio on the crop is a failure to reject that input. Body-shape classification requires separately labelled examples.

```python
from pathlib import Path
import sys

# Start Jupyter in the project folder or its notebooks folder.
root = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "backend/catalogue.json").is_file()), None)
if root is None:
    raise RuntimeError("Open the project directory first. In Colab, upload/mount the project and %cd into it.")
sys.path.insert(0, str(root / "backend"))
from PIL import Image, ImageOps
from body_pose import measure_body_pose

photo = root / "generated_images/exec-e90d3302-b016-49b6-b4fd-fb355ffefe05.png"
with Image.open(photo) as source:
    full = ImageOps.exif_transpose(source).convert("RGB")
# For a different photo, adjust the crop after checking its preview.
cropped = full.crop((0, 0, full.width, int(full.height * 0.38)))
```

```python
for label, image in [("Full body", full), ("Above-hip crop", cropped)]:
    display(image.resize((round(image.width * 230 / image.height), 230)))
    ratio = measure_body_pose(image)
    expected_ratio = label == "Full body"
    print(label, "| shoulder-to-hip ratio:", ratio, "|", "PASS" if (ratio is not None) == expected_ratio else "CHECK")
```
